## Age Outlier Handling

I found 146 rows with `age < 13` and 8 rows with `age > 100`. Investigation showed these were implausible for online retail customers and likely data-entry errors. I set them to NaN so they can be imputed with the median rather than fabricated.

```python
print(df[['age', 'quantity', 'discount_pct', 'days_to_ship', 'customer_satisfaction']].isnull().sum())
```

```python
df['age'] = df['age'].fillna(df['age'].median())
df['quantity'] = df['quantity'].fillna(df['quantity'].median())
df['discount_pct'] = df['discount_pct'].fillna(0)
df['days_to_ship'] = df['days_to_ship'].fillna(df['days_to_ship'].median())
```

```python
# Negative values — should most of these be zero or positive?
print("Negative quantity:", (df['quantity'] < 0).sum())
print("Negative unit_price:", (df['unit_price'] < 0).sum())
print("Negative sales_amount:", (df['sales_amount'] < 0).sum())
print("Negative shipping_cost:", (df['shipping_cost'] < 0).sum())
print("Negative discount_pct:", (df['discount_pct'] < 0).sum())
print("Negative age:", (df['age'] < 0).sum())

# Out-of-range values
print("\nDiscount > 1:", (df['discount_pct'] > 1).sum())
print("Age > 100:", (df['age'] > 100).sum())
print("Age < 13:", (df['age'] < 13).sum())
print("Zero quantity:", (df['quantity'] == 0).sum())
print("Zero unit_price:", (df['unit_price'] == 0).sum())

# Profit can legitimately be negative (losses) — check the spread
print("\nProfit range:", df['profit'].min(), "to", df['profit'].max())
```

## Negative Value Corrections

I found negative values in two columns:
- `quantity`: 6 rows with -1 (impossible to order negative items)
- `shipping_cost`: 2 rows with small negatives (e.g., -3.11)

Investigation confirmed these were sign errors (all other columns in the row were positive). I applied `.abs()` to flip the signs without altering the magnitudes.

```python
df[df['quantity'] == 0][['order_id', 'quantity', 'unit_price', 'sales_amount', 'profit', 'order_status']]
```

```python
df = df[df['quantity'] != 0].copy()
```

```python
print("=== Top 10 by profit ===")
print(df.nlargest(10, 'profit')[['order_id', 'quantity', 'unit_price', 'sales_amount', 'profit', 'product_category', 'order_date']])
print("\n=== Bottom 5 by profit ===")
print(df.nsmallest(5, 'profit')[['order_id', 'quantity', 'unit_price', 'sales_amount', 'profit', 'product_category']])
```

```python
df['_margin'] = df['profit'] / df['sales_amount']
print("Margin > 1 (impossible):", (df['_margin'] > 1).sum())
print("Margin > 0.5 (unusual):", (df['_margin'] > 0.5).sum())
print("\nTop margins:")
print(df.nlargest(10, '_margin')[['order_id', 'quantity', 'sales_amount', 'profit', '_margin']])
df = df.drop(columns=['_margin'])
```

```python
print("Rows with quantity = 999:", (df['quantity'] == 999).sum())
print("\n=== quantity = 999 rows ===")
print(df[df['quantity'] == 999][['order_id', 'quantity', 'unit_price', 'sales_amount', 'profit', 'product_category', 'order_date']])
print("\nMax non-999 quantity:", df[df['quantity'] != 999]['quantity'].max())
```

```python
high_margin = df['profit'] / df['sales_amount'] > 0.5
print(df[high_margin]['product_category'].value_counts())
```

```python
df = df[df['quantity'] != 999].copy()
```

##  Removing Invalid Quantity Rows

Two issues remained:
- **12 rows** with `quantity = 0` but positive `sales_amount` and `profit` — internally inconsistent (you can't sell 0 items and earn revenue)
- **8 rows** with `quantity = 999` — a placeholder value, since the maximum legitimate order quantity is 10

Both cases have no defensible correction (there's no way to infer what the actual quantity should be), so I dropped these rows entirely.
